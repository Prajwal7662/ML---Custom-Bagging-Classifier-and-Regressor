```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
```

```python
x,y = make_classification(n_samples = 1000,n_classes=2,n_features=20,random_state=1)
```

```python
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.30,random_state=1)
```

```python
x_train
```

```python
y_train
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
```

```python
lr_clf = LogisticRegression()
dt_clf = DecisionTreeClassifier()
sv_clf = SVC()
gn_clf = GaussianNB()
```

```python
# for ensembles

from sklearn.ensemble import VotingClassifier
```

```python
 ensemble_clf = VotingClassifier(estimators=[('lr',lr_clf),('dt',dt_clf),('sv',sv_clf),('gn',gn_clf)],voting='hard')
```

```python
ensemble_clf
```

```python
ensemble_clf.fit(x_train,y_train)
```

```python
y_pred = ensemble_clf.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
y_pred
```

#Bagging Regressor


```python
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
```

```python
x,y = make_regression(n_samples=1000,n_features=10,noise=0.1,random_state=1)
```

```python
x
```

```python
y
```

```python
# train test split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.30,random_state=1)
```

```python
x_train
```

```python
x_test
```

```python
y_train
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
```

```python
lr = LinearRegression()
dtr = DecisionTreeRegressor()
svr = SVR(kernel='linear')
```

```python
# to create ensemble

from sklearn.ensemble import VotingRegressor
ensemble_regressor = VotingRegressor(estimators=[('lr',lr),('dtr',dtr),('svr',svr)])
```

```python
ensemble_regressor
```

```python
ensemble_regressor.fit(x_train,y_train)
```

```python
y_pred = ensemble_regressor.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test,y_pred)
```
